### Profiling customers dataset

```python
# Reading the data — zip prefix as string so leading zeros survive
olist_customers_df = pd.read_csv('../data/raw/olist_customers_dataset.csv',dtype={'customer_zip_code_prefix': str})
olist_customers_df.head(5)
```

```python
# Displaying the information about the dataframe
olist_customers_df.info()
```

```python
# Displaying the summary statistics of the dataframe
olist_customers_df.describe()
```

```python
# Displaying the null values in the dataframe
olist_customers_df.isnull().sum()
```

```python
# Displaying the number of unique values in each column of the dataframe
olist_customers_df.nunique()
```

```python
# Displaying zip code prefix counts 
zip_count = olist_customers_df['customer_zip_code_prefix'].str.len().value_counts()
zip_count
```

```python
# Displaying repeated customers in the dataframe
count = olist_customers_df['customer_unique_id'].value_counts()
repeated_customers = count[count > 1]
repeated_customers
```

```python
# Compare the state codes in each table, both directions
seller_states = set(olist_sellers_df['seller_state'].unique())
customer_states = set(olist_customers_df['customer_state'].unique())

print("In customers, not in sellers:", sorted(customer_states - seller_states))
print("In sellers, not in customers:", sorted(seller_states - customer_states))
print("Union:", len(seller_states | customer_states))
```

### Profiling orders dataset

```python
# Reading the data — parse dates for the order purchase, approval, delivery, and estimated delivery dates
olist_orders_df = pd.read_csv('../data/raw/olist_orders_dataset.csv',parse_dates=[3,4,5,6,7])
olist_orders_df.head()
```

```python
# Displaying the information about the dataframe(verbose=True for more than 100 columns)
olist_orders_df.info(verbose=True)
```

```python
# Displaying the summary statistics of the dataframe
olist_orders_df.describe()
```

```python
# Displaying the null values in the dataframe
olist_orders_df.isnull().sum()
```

```python
# Displaying the number of unique values in each column of the dataframe
olist_orders_df.nunique()
```

```python
# Displaying the value counts of the 'order_status' column in the dataframe
olist_orders_df['order_status'].value_counts()
```

```python
olist_orders_df.isnull().sum()
```

```python
# Displaying the rows with null values in the 'order_approved_at' column of the dataframe
olist_orders_df[olist_orders_df['order_approved_at'].isnull()]
```

```python
# Displaying the rows with null values in the 'order_delivered_carrier_date' column of the dataframe
approved_null = olist_orders_df[(olist_orders_df['order_approved_at'].isnull())]
approved_null['order_status'].value_counts()
```

```python
# Displaying order status and delivery dates for delivered orders with null carrier delivery date
olist_orders_df[(olist_orders_df['order_status']=='delivered') & (olist_orders_df['order_approved_at'].isnull())]
```

```python
# Displaying the rows with null values in the 'order_delivered_carrier_date' column of the dataframe
olist_orders_df[olist_orders_df['order_delivered_carrier_date'].isnull()]
```

```python
# Displaying the rows where the 'order_delivered_carrier_date' column is null
carrier_null = olist_orders_df[olist_orders_df['order_delivered_carrier_date'].isnull()]
carrier_null['order_status'].value_counts()
```

```python
# Displaying order status for canceled orders and delivery dates for delivered orders with null carrier delivery date
olist_orders_df[(olist_orders_df['order_status']=='canceled') &
                (olist_orders_df['order_delivered_customer_date'].notnull())]
```

```python
olist_orders_df[olist_orders_df['order_delivered_customer_date'].isnull()]
```

```python
delivery_null = olist_orders_df[olist_orders_df['order_delivered_customer_date'].isnull()]
delivery_null['order_status'].value_counts()
```

```python
# Compare the customer ids in each table, both directions
customers = set(olist_customers_df['customer_id'].unique())
orders = set(olist_orders_df['customer_id'].unique())

print('Present in customers but not in orders:', sorted(customers - orders))
print('Present in orders but not in customers:', sorted(orders - customers))
print('Union:', len(customers | orders))
```

```python
d = (
    ((olist_orders_df['order_status'] == 'delivered') & olist_orders_df['order_approved_at'].isnull())
    | ((olist_orders_df['order_status'] == 'delivered') & olist_orders_df['order_delivered_carrier_date'].isnull())
    | ((olist_orders_df['order_status'] == 'delivered') & olist_orders_df['order_delivered_customer_date'].isnull())
    | ((olist_orders_df['order_status'] == 'canceled') & olist_orders_df['order_delivered_customer_date'].notnull())
)
olist_orders_df[d]['order_id'].nunique()
```
