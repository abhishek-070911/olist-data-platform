```python
# Difference per order (payments minus items), rounded to cents so tiny float errors don't count
totals['diff'] = (totals['payment_total'] - totals['item_total']).round(2)

# Label each order by how its two totals compare
totals['result'] = 'only in one table'
totals.loc[totals['diff'] == 0, 'result'] = 'equal'
totals.loc[totals['diff'].abs() == 0.01, 'result'] = 'off by 1 cent'
totals.loc[totals['diff'] > 0.01, 'result'] = 'payments more'
totals.loc[totals['diff'] < -0.01, 'result'] = 'payments less'

print(totals['result'].value_counts())
totals.groupby('result')['diff'].describe()
```

```python
# Are the copies of each repeated review_id identical, apart from order_id?
repeated_reviews = olist_reviews_df[olist_reviews_df['review_id'].isin(repeated_review_ids.index)]

content_cols = ['review_score', 'review_comment_title', 'review_comment_message',
                'review_creation_date', 'review_answer_timestamp']

# One row per repeated review_id; each number = how many different values that column has across its copies
values_per_review = repeated_reviews.groupby('review_id')[content_cols].nunique(dropna=False)
print('Repeated review_ids checked:', len(values_per_review))

# Per column: how many review_ids have copies that disagree (0 everywhere = identical copies)
(values_per_review > 1).sum()
```

```python
# How many reviews each order has (1, 2 or 3)
orders_reviews = olist_reviews_df.groupby('order_id')['review_id'].agg(review_id_count = 'count')
orders_reviews.value_counts()
```

```python
# Orders with more than one review: do their reviews differ?
reviews_per_order = olist_reviews_df['order_id'].value_counts()
multi_review_orders = reviews_per_order[reviews_per_order>1]

multi_reviews = olist_reviews_df[olist_reviews_df['order_id'].isin(multi_review_orders.index)]
print('Orders with more than one review:', len(multi_review_orders))
print('Reviews rows for those orders:', len(multi_reviews))

# One row per order; each number = how many different values that column has across its reviews
values_per_order = multi_reviews.groupby('order_id')[content_cols].nunique(dropna=False)

# Per column: how many orders have reviews that disagree
(values_per_order > 1).sum()
```

```python
# For orders whose reviews give different scores: how far apart are the lowest and highest score?
score_range = multi_reviews.groupby('order_id')['review_score'].agg(['min', 'max'])
score_range = score_range[score_range['min'] != score_range['max']]
print('Orders whose reviews give different scores:', len(score_range))
(score_range['max'] - score_range['min']).value_counts().sort_index().rename_axis('score_gap').rename('orders')
```

```python
# Counting reviews by title and message: only one, both, or neither
only_titles = olist_reviews_df[olist_reviews_df['review_comment_message'].isnull() & olist_reviews_df['review_comment_title'].notnull()]
only_messages = olist_reviews_df[olist_reviews_df['review_comment_message'].notnull() & olist_reviews_df['review_comment_title'].isnull()]
titles_and_messages = olist_reviews_df[olist_reviews_df['review_comment_message'].notnull() & olist_reviews_df['review_comment_title'].notnull()]
no_titles_no_messages = olist_reviews_df[olist_reviews_df['review_comment_message'].isnull() & olist_reviews_df['review_comment_title'].isnull()]
print('Reviews with only titles :',len(only_titles))
print('Reviews with only messages :',len(only_messages))
print('Reviews with both messages and titles :',len(titles_and_messages))
print('Reviews with no messages and no titles :',len(no_titles_no_messages))
```

```python
# Bringing each review's order status and dates next to it
reviews_with_orders = pd.merge(olist_reviews_df,olist_orders_df[['order_id','order_status','order_purchase_timestamp','order_delivered_customer_date','order_estimated_delivery_date']],
         on = 'order_id',how = 'left')

print('Rows after merge:', len(reviews_with_orders))
```

```python
# Reviews created before their order was bought (compare dates only; purchase has a time part)
purchase_day = reviews_with_orders['order_purchase_timestamp'].dt.normalize()
before_purchase = reviews_with_orders[reviews_with_orders['review_creation_date'] < purchase_day]
print('Reviews created before the order was bought:', len(before_purchase))
before_purchase[['review_id', 'order_id', 'order_status', 'order_purchase_timestamp', 'review_creation_date']]
```

```python
# Order status of the 64 reviews created before their order was bought
purchase_before_status = before_purchase['order_status'].value_counts()
purchase_before_status
```

```python
# How many days before the purchase those 64 reviews were created
purchase_before = before_purchase['order_purchase_timestamp'].dt.normalize()
early_days = purchase_before - before_purchase['review_creation_date']
days = early_days.dt.days
days.describe()
```

```python
# Reviews created before the order reached the customer
creation_day = reviews_with_orders['review_creation_date'].dt.normalize()
delivered_day = reviews_with_orders['order_delivered_customer_date'].dt.normalize()

no_delivery_date = reviews_with_orders['order_delivered_customer_date'].isnull()
before_delivery = creation_day < delivered_day 

print('Reviews whose order has no delivery date  :', no_delivery_date.sum())
print('Reviews created before the order arrived  :', before_delivery.sum())
```

```python
# Order status of those two groups
print('No delivery date:')
print(reviews_with_orders.loc[no_delivery_date, 'order_status'].value_counts())
print()
print('Created before delivery:')
print(reviews_with_orders.loc[before_delivery, 'order_status'].value_counts())
```

```python
# For those early reviews, days between the estimated delivery date and the review's creation date
estimated_day = reviews_with_orders['order_estimated_delivery_date'].dt.normalize()
early_reviews = before_delivery | no_delivery_date
days_after_estimate = (creation_day - estimated_day)[early_reviews].dt.days
print('Early reviews:', early_reviews.sum())
days_after_estimate.value_counts().head(10)
```
