### Profiling items dataset

```python
# Reading the data — parse dates for the shipping_limit_date
olist_items_df = pd.read_csv('../data/raw/olist_order_items_dataset.csv',parse_dates=['shipping_limit_date'])
olist_items_df[:5]
```

```python
# Displaying the information about the dataframe(verbose=True for more than 100 columns)
olist_items_df.info(verbose=True)
```

```python
# Displaying the null values in the dataframe
olist_items_df.isnull().sum()
```

```python
# Displaying the number of unique values in each column of the dataframe
olist_items_df.nunique()
```

```python
# Displaying the summary statistics of the dataframe
olist_items_df.describe(include='all')
```

```python
# Count duplicate (order_id, order_item_id) pairs — 0 means the pair is unique
olist_items_df.duplicated(subset = ['order_id', 'order_item_id']).sum()
```

```python
# Within each order, does order_item_id run 1, 2, 3... with no gaps?
seq = olist_items_df.groupby('order_id')['order_item_id'].agg(['min', 'max', 'count'])
print('Orders checked:', len(seq))

gaps = seq[(seq['min'] != 1) | (seq['max'] != seq['count'])]
print('Orders with a gap or not starting at 1:', len(gaps))
gaps.head()
```

```python
# Distinct order_id in orders vs items
unique_orders = olist_orders_df['order_id'].nunique()
unique_items = olist_items_df['order_id'].nunique()

print('unique_orders :',unique_orders)
print('unique_items :',unique_items)

print()

print('Difference in order_ids :',unique_orders - unique_items)
```

```python
# Merging the olist_orders_df and olist_items_df dataframes on 'order_id' using a left join to get all 
# orders and their corresponding items, if any. This will help identify orders that do not have associated items in 
# the olist_items_df.
merge_orders_items = pd.merge(olist_orders_df,olist_items_df,on = 'order_id',how = 'left')
merge_orders_items_nulls = merge_orders_items[merge_orders_items['order_item_id'].isnull()]
merge_orders_items_nulls
```

```python
# Checking the order_status value counts for merged orders with null items and all orders in olist_orders_df
print('Merged orders:', merge_orders_items_nulls['order_status'].value_counts())
print()
print('All orders:', olist_orders_df['order_status'].value_counts())
```

```python
# Checking for null values in the merged dataframe with null items
merge_orders_items_nulls.isnull().sum()   
```

```python
# Looking for order_status = shipped,invoiced,created in the merged dataframe with null items
merge_orders_items_nulls[merge_orders_items_nulls['order_status'].isin(['shipped', 'invoiced','created'])]
```

```python
# Summing the duplicated combinations of 'order_id' and 'product_id' in the dataframe
olist_items_df.duplicated(subset=['order_id', 'product_id']).sum()
```

```python
# Rows of repeated (order_id, product_id) pairs, and how many orders they're in
repeats = olist_items_df[olist_items_df.duplicated(subset=['order_id', 'product_id'], keep=False)]
repeats['order_id'].nunique()
```

```python
# All rows of the repeated (order_id, product_id) pairs, sorted by order
repeats.sort_values(['order_id', 'order_item_id'])
```

```python
# Within each repeated (order_id, product_id) pair, count distinct values per column
pair = repeats.groupby(['order_id','product_id'])[['seller_id','shipping_limit_date','price','freight_value']].nunique()
print('Repeated pairs',len(pair))
(pair>1).sum()
```

```python
# Highest distinct count per column — 1 everywhere means no pair ever differs
pair.max()
```

```python
# Do order_items link cleanly to orders and sellers? Checked in both directions
items_orders  = set(olist_items_df['order_id'])
all_orders    = set(olist_orders_df['order_id'])
items_sellers = set(olist_items_df['seller_id'])
all_sellers   = set(olist_sellers_df['seller_id'])

print('order_id in items but not in orders :', len(items_orders - all_orders))
print('order_id in orders but not in items :', len(all_orders - items_orders))
print('seller_id in items but not in sellers:', len(items_sellers - all_sellers))
print('seller_id in sellers but not in items:', len(all_sellers - items_sellers))
```

```python
# How many orders have items from more than one seller?
sellers_per_order = olist_items_df.groupby('order_id')['seller_id'].nunique()
print('Orders with items:', len(sellers_per_order))
print('Orders with more than one seller:', (sellers_per_order > 1).sum())
sellers_per_order.value_counts().sort_index()
```

```python
# How many item rows have zero freight, in how many orders and sellers?
zero_freight = olist_items_df[olist_items_df['freight_value'] == 0]
print('Rows with zero freight:', len(zero_freight))
print('Orders with a zero-freight row:', zero_freight['order_id'].nunique())
print('Sellers with a zero-freight row:', zero_freight['seller_id'].nunique())
print('Share of all rows:', round(len(zero_freight) / len(olist_items_df) * 100, 2), '%')
```
